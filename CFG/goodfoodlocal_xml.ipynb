```python

import xml.etree.ElementTree as ET
import pandas as pd

# 1. Load your XML (make sure 'capitalgrowth-london.xml' is in your notebook dir)
tree = ET.parse('../myenv/capstone data/good-food-local-ap-2025_table_xml.php.xml')
root = tree.getroot()

# 2. Build a list of dicts, one per <marker>
records = []
for m in root.findall('marker'):
    # Grab every attribute on this element
    attrs = m.attrib.copy()
    records.append(attrs)

# 3. Turn into a DataFrame
df = pd.DataFrame(records)

# 4. Quick sanity‐check
print(df.shape, "markers loaded")
df
```

```python
df = df.rename(columns={
        "section1_score": 'Food governance & strategy' , "section2_score": 'Community food growing', "section3_score": 'Community food action', "section14_score": 'Stemming the flood of unhealthy food', "section15_score": 'Sustainable food economy', "section16_score": 'Catering & procurement', "section17_score": 'Food for the planet', "section99_score": 'Ethnic and cultural diversity in the food system'
    })
df
```
