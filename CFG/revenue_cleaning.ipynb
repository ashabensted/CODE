```python

def extract_theme(var):
    parts = var.split("_")
    if len(parts) >= 3:
        return parts[1]   # middle part is theme
    elif len(parts) == 2:
        return parts[1]   # if only 2 parts, treat second as theme
    else:
        return None

dict['Theme'] = dict['Data item name'].apply(extract_theme)
dict['Theme'].value_counts()
```

```python
import re
RULES = [
    # --- Service areas ---
    (r"^edu",                   "Education"),
    (r"^csc",                   "Children’s Social Care"),
    (r"^asc",                   "Adult Social Care"),
    (r"^ph",                    "Public Health"),
    (r"^hous",                  "Housing"),
    (r"^hra",                   "Housing Revenue Account"),
    (r"^cul",                   "Culture"),
    (r"^env",                   "Environment"),
    (r"^plan",                  "Planning & Development"),
    (r"^culenvplan",           "Culture / Environment / Planning"),  # if mixed prefix
    (r"^pol",                   "Police"),
    (r"^frs",                   "Fire & Rescue"),
    (r"^cen",                   "Central Services"),
    (r"^trans",                 "Transport"),
    (r"^grant",                 "Grants"),
    (r"^totsx",                 "Total Service Expenditure"),

    # --- Financing & reserves ---
    (r"^levy",                  "Levies"),
    (r"^res",                   "Reserves"),
    (r"^app",                   "Appropriations"),
    (r"^adj",                   "Adjustments"),
    (r"^cap",                   "Capital"),
    (r"^prov",                  "Provisions"),
    (r"^pyadj",                 "Prior Year Adjustments"),

    # --- Income & trading ---
    (r"^inc",                   "Income"),
    (r"^int",                   "Interest"),
    (r"^invprp",                "Investment Properties"),
    (r"^trad",                  "Trading"),
    (r"^hb",                    "Housing Benefits"),
    (r"^brsupp",                "Business Rates Support"),
    (r"^netrevexp",             "Net Revenue Expenditure"),
    (r"^netcurrtot",           "Net Current Expenditure"),

    # --- Special ---
    (r"^cv19",                  "COVID-19"),
    (r"^dsb",                   "Disbursements"),
    (r"^equal",                 "Equal Pay"),
    (r"^sub",                   "Subsidies"),
    (r"^sfc",                   "Schools Funding"),
    (r"^cera",                  "Collection Fund Adjustments"),
]

def to_theme_group(theme: str) -> str:
    if not isinstance(theme, str):
        return "Other / Unmapped"
    t = theme.lower()
    for pat, label in RULES:
        if re.match(pat, t, flags=re.IGNORECASE):
            return label
    return "Other / Unmapped"

dict["Theme_group"] = dict["Theme"].apply(to_theme_group)
```

```python
def norm_code(s: str) -> str:
    if not isinstance(s, str):
        return ""
    s = s.strip()
    # match dd's hyphenated codes to underscore-coded_df columns
    return re.sub(r"[^0-9a-zA-Z_]+", "", s.replace("-", "_")).lower()

# Build normalized lookup maps from df
df_norm = df.copy()
df_norm["code_norm"] = df_norm["Data item name"].astype(str).map(norm_code)

sub_theme_map = df_norm.set_index("code_norm")["Theme"].to_dict()
theme_map     = df_norm.set_index("code_norm")["Theme_group"].to_dict()

# Prepare column arrays
orig_cols  = list(rev.columns)
norm_cols  = [norm_code(c) for c in orig_cols]
themes     = [theme_map.get(nc, "") for nc in norm_cols]      # top-level
subthemes  = [sub_theme_map.get(nc, "") for nc in norm_cols]  # second-level

# Apply MultiIndex columns
multi = rev.copy()
multi.columns = pd.MultiIndex.from_arrays(
    [themes, subthemes, orig_cols],
    names=["theme", "sub_theme", "Variable"]
)

# Quick coverage check
matched = sum(bool(t or st) for t, st in zip(themes, subthemes))
print(f"Matched metadata for {matched} / {len(orig_cols)} columns.")
```

```python
def _norm_code(s: str) -> str:
    s = str(s).strip().replace("-", "_")
    s = re.sub(r"[^0-9A-Za-z_]", "", s)
    return s.lower()

# 1) build code -> description mapping from dd
md = df.loc[:, ["Data item name", "Description"]].dropna(subset=["Data item name"]).copy()
md["code_norm"] = md["Data item name"].map(_norm_code)
code2desc = md.set_index("code_norm")["Description"].to_dict()

# 2) find which level in df_multi is the 'code' level
col_names = list(multi.columns.names)
try:
    code_level = col_names.index("code")
except ValueError:
    # if not explicitly named, assume last level is the code
    code_level = len(col_names) - 1

# 3) build new tuples with description inserted just before code
new_tuples = []
matched = 0
for tpl in multi.columns.to_flat_index():
    tpl = tpl if isinstance(tpl, tuple) else (tpl,)
    code = tpl[code_level]
    desc = code2desc.get(_norm_code(code), "")
    if desc:
        matched += 1
    new_tpl = tuple(list(tpl[:code_level]) + [desc] + list(tpl[code_level:]))
    new_tuples.append(new_tpl)

# 4) assign the new MultiIndex (… , 'description', 'code')
new_names = list(col_names)
new_names.insert(code_level, "description")

multi_desc = multi.copy()
multi_desc.columns = pd.MultiIndex.from_tuples(new_tuples, names=new_names)

print(f"Added descriptions for {matched} / {len(multi.columns)} columns.")
print("Column levels now:", multi_desc.columns.names)
```
