```python

import xml.etree.ElementTree as ET
import pandas as pd

# 1. Load your XML (make sure 'capitalgrowth-london.xml' is in your notebook dir)
tree = ET.parse('../myenv/capstone data/capitalgrowth-london.xml')
root = tree.getroot()

# 2. Build a list of dicts, one per <marker>
records = []
for m in root.findall('marker'):
    # Grab every attribute on this element
    attrs = m.attrib.copy()
    # Convert lat/lng to floats
    if 'lat' in attrs:  attrs['lat'] = float(attrs['lat'])
    if 'lng' in attrs:  attrs['lng'] = float(attrs['lng'])
    records.append(attrs)

# 3. Turn into a DataFrame
df = pd.DataFrame(records)

# 4. Quick sanity‐check
print(df.shape, "markers loaded")
df.head(5)
```

Project descrition only available for first 10, 
The reason you only see a handful of non‐blank project_decs in the XML dump is that the map endpoint is deliberately kept “light” – it only emits a short blurb (or whatever legacy snippet exists) for a handful of spaces. The full write-ups you see on the site are loaded on demand when you click through to a space’s Detail page (via that fancy iframe /processors/mapEntry.php?space_id=… call), not in the bulk XML feed.

```python
import requests
from bs4 import BeautifulSoup

def fetch_space_details(space_id):
    """
    Given a space_id, load the mapEntry iframe and scrape out:
      - full_description
      - run_by
      - campaign_name, campaign_url
      - website_url
      - social_links: list of {platform, url}
    """
    url = f"https://www.capitalgrowth.org/processors/mapEntry.php?space_id={space_id}&area_url=london"
    resp = requests.get(url)
    resp.raise_for_status()
    soup = BeautifulSoup(resp.text, "html.parser")
    
    # 1) Full description = first <p class="bot15"> (but *not* the one with "Run by")
    desc_p = soup.find_all("p", class_="bot15")
    full_desc = ""
    for p in desc_p:
        # skip the one that has <strong>Run by</strong>
        if p.find("strong", string=lambda s: s and "Run by" in s):
            continue
        full_desc = p.get_text(" ", strip=True)
        break


    # 3) Campaign panel (<p class="gpanel">)
    camp_p = soup.find("p", class_="gpanel")
    campaign_name = campaign_url = ""
    if camp_p:
        a = camp_p.find("a", href=True)
        if a:
            campaign_name = a.get_text(strip=True)

    # 4) Website button
    site_btn = soup.find("a", class_="btn btn-default", href=True)
    website_url = site_btn["href"] if site_btn else ""

       # 5) Social links in the Location panel
    social_links = []
    loc_block = soup.find("div", class_="col-sm-7 bot25 borderTop0")
    if loc_block:
        # look at each <p class="bot25"> under that div
        for p in loc_block.find_all("p", class_="bot25"):
            for a in p.find_all("a", href=True):
                href = a["href"]
                text = a.get_text(strip=True)
                # infer platform
                if "twitter.com" in href:
                    platform = "Twitter"
                elif "facebook.com" in href:
                    platform = "Facebook"
                else:
                    # fallback to link text or hostname
                    platform = text or href.split("//")[-1].split("/")[0]
                social_links.append({"platform": platform, "url": href})

    return {
        "space_id":      space_id,
        "full_desc":     full_desc,
        "campaign_name": campaign_name,
        "website_url":   website_url,
        "social_links":  social_links
    }



```

```python
from tqdm import tqdm

# fetch into a list of dicts
all_details = []
for sid in tqdm(df["space_id"].unique()):
    try:
        all_details.append(fetch_space_details(sid))
    except Exception as e:
        print("failed", sid, e)

# build a small DataFrame and join
details_df = pd.DataFrame(all_details).set_index("space_id")
details_df.head(50)
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# building one “document” per space_type by concatenating name+description
grouped = (
    df
    .fillna({'full_desc': ''})
    .groupby('space_type')[['space_name','full_desc']]
    .apply(lambda g: " ".join((g['space_name'] + " " + g['full_desc']).tolist()))
    # result is a Series indexed by space_type
)

# running tf-idf on this Series
vectorizer = TfidfVectorizer(ngram_range=(1,3),
    max_df=0.99,      # ignore very common words
    min_df=1,        # ignore super-rare words
    stop_words='english'
)
tfidf_matrix = vectorizer.fit_transform(grouped.values)
terms = vectorizer.get_feature_names_out()

# tf-idf dataframe
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    index=grouped.index,
    columns=terms
)

# showing top unigrams, bigrams, and trigrams for each space_type
top_n = 10
for space_type, row in tfidf_df.iterrows():
    top_terms = row.nlargest(top_n).index.tolist()
    print(f"Type {space_type}: {', '.join(top_terms)}")

```

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_selection import chi2, SelectKBest

# 1) build document‐term matrix for each space 
cv = CountVectorizer(stop_words='english', ngram_range=(1,2))
X = cv.fit_transform(df['full_desc'].fillna(''))
y = df['space_type']

# 2) for each type vs. rest, pick top k features 
top_features = {}
for t in df['space_type'].unique():
    chi2_scores, p_vals = chi2(X, (y == t).astype(int))
    idxs = np.argsort(chi2_scores)[-10:]
    top_features[t] = [cv.get_feature_names_out()[i] for i in idxs]
```

```python


# turn into a DataFrame: rows are types, columns are the top k features
df_feats = pd.DataFrame.from_dict(
    top_features, 
    orient='index', 
    columns=[f"top{i+1}" for i in range(len(next(iter(top_features.values()))))]
)
df_pretty = df_feats.T
df_pretty.index.name = "feature_rank"
print(df_pretty)

```

```python
import re
import html
import string

from nltk.tokenize import word_tokenize
from nltk.corpus   import stopwords
from nltk.stem     import WordNetLemmatizer

# Pre-compile your cleaner objects
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
punct_trans = str.maketrans(string.punctuation, ' ' * len(string.punctuation))

def clean_and_lemmatize(text: str) -> str:
    # 1) strip HTML and unescape entities
    text = re.sub(r'<[^>]+>', ' ', text)
    text = html.unescape(text)

    # 2) lowercase & remove punctuation
    text = text.lower().translate(punct_trans)

    # 3) remove stray numbers
    text = re.sub(r'\b[\d]+\b', ' ', text)

    # 4) tokenize, remove stopwords and non-alpha, then lemmatize
    tokens = word_tokenize(text)
    clean_tokens = []
    for tok in tokens:
        if tok.isalpha() and tok not in stop_words:
            lemma = lemmatizer.lemmatize(tok)
            clean_tokens.append(lemma)
    return " ".join(clean_tokens)

# Apply to your DataFrame
df['corpus'] = (df['space_name'].fillna('') + ' ' + df['full_desc'].fillna(''))
df['corpus'] = df['corpus'].apply(clean_and_lemmatize)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1) per-space TF-IDF with binary term presence
tfidf = TfidfVectorizer(
    stop_words='english',
    ngram_range=(1,3),
    binary=True,
    norm='l2'  # default
)
X = tfidf.fit_transform(df['corpus'].fillna(''))
terms = tfidf.get_feature_names_out()

# 2) make a DataFrame
tfidf_df = pd.DataFrame(X.toarray(), index=df.index, columns=terms)

# 3) average TF-IDF by space_type
avg_by_type = tfidf_df.join(df['space_type']).groupby('space_type').mean()

# 4) see top terms per type
for t, row in avg_by_type.iterrows():
    top10 = row.nlargest(10).index.tolist()
    print(f"Type {t}: {', '.join(top10)}")
```
